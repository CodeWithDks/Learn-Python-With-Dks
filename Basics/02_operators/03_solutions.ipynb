## Setup

```python
import math

def check(fn, cases, label=None):
    """Run fn on every (args_tuple, expected) pair and report. Never crashes the notebook."""
    label = label or fn.__name__
    failed = 0
    for args, expected in cases:
        try:
            got = fn(*args)
        except Exception as e:
            got = f"raised {type(e).__name__}: {e}"
        same = (math.isclose(got, expected, abs_tol=1e-9)
                if isinstance(got, float) and isinstance(expected, float) else got == expected)
        if not same:
            failed += 1
            print(f"  ❌ {label}{args}\n       expected: {expected!r}\n       got     : {got!r}")
    if failed:
        print(f"❌ {label}: {failed}/{len(cases)} checks failed - keep going")
    else:
        print(f"✅ {label}: all {len(cases)} checks passed")
```

### Q1 · Days → months + days

Return `(months, days)` for a number of days, counting every month as 30 days.

`days_to_months_days(475) → (15, 25)`

```python
def days_to_months_days(days):
    return divmod(days, 30)
```

```python
check(days_to_months_days, [((475,), (15, 25)), ((0,), (0, 0)), ((90,), (3, 0)), ((29,), (0, 29))])
```

### Q2 · Leap year

Return `True` for a leap year: divisible by 4, **except** centuries, which must be divisible by 400.

`is_leap(2024) → True`   `is_leap(1900) → False`   `is_leap(2000) → True`

```python
def is_leap(year):
    return year % 4 == 0 and (year % 100 != 0 or year % 400 == 0)
```

```python
check(is_leap, [((2024,), True), ((1900,), False), ((2000,), True), ((2023,), False), ((2100,), False)])
```

### Q3 · Gate pass

A person may enter if they are **at least 16**, **have a membership card**, and are **not banned**.

`can_enter(20, True, False) → True`

```python
def can_enter(age, has_card, is_banned):
    return age >= 16 and has_card and not is_banned
```

```python
check(can_enter, [((20, True, False), True), ((15, True, False), False), ((20, False, False), False), ((20, True, True), False), ((16, True, False), True)])
```

### Q4 · Peak hours (chained comparison)

Peak hours are **06:00–09:59** and **17:00–20:59**. Given an hour (0–23), return `True` if it's peak.

`in_peak_hours(7) → True`   `in_peak_hours(10) → False`

```python
def in_peak_hours(hour):
    return 6 <= hour < 10 or 17 <= hour < 21
```

```python
check(in_peak_hours, [((6,), True), ((9,), True), ((10,), False), ((16,), False), ((17,), True), ((20,), True), ((21,), False), ((0,), False)])
```

### Q5 · Permission check (bitwise)

Permissions: `GYM=1`, `POOL=2`, `TRAINER=4`, `STEAM=8`. Return `True` if `flags` includes **all** the bits in `perm`.

`has_permission(5, 1) → True` (5 = GYM+TRAINER)   `has_permission(5, 2) → False`

```python
def has_permission(flags, perm):
    return (flags & perm) == perm
```

```python
check(has_permission, [((5, 1), True), ((5, 2), False), ((7, 4), True), ((7, 2 | 8), False), ((15, 2 | 8), True), ((0, 1), False)])
```

### Q6 · Grant and revoke

Return the new flags after **granting** `perm` (switch the bit on) — and in a second function, after **revoking** it (switch it off). Granting twice or revoking something you don't have must not break anything.

`grant(5, 2) → 7`   `revoke(7, 4) → 3`

```python
def grant(flags, perm):
    return flags | perm

def revoke(flags, perm):
    return flags & ~perm
```

```python
check(grant,  [((5, 2), 7), ((7, 2), 7), ((0, 8), 8)])
check(revoke, [((7, 4), 3), ((3, 4), 3), ((15, 15), 0)])
```

### Q7 · Rupees → paise (the float trap)

Convert a rupee amount (float) to **whole paise** (int). Naive `int(rupees * 100)` gets some of these wrong — find out which, and fix it.

`rupees_to_paise(19.99) → 1999`   `rupees_to_paise(0.29) → 29`

```python
def rupees_to_paise(rupees):
    return round(rupees * 100)
```

```python
check(rupees_to_paise, [((19.99,), 1999), ((0.29,), 29), ((1499.5,), 149950), ((0.1 + 0.2,), 30), ((0,), 0)])
```

### Q8 · Late fine with grace and cap

Fine in **rupees**: first **3 days late are free**, then **₹10 per day**, never more than **₹500**. Negative or zero days → 0.

`late_fine(2) → 0`   `late_fine(9) → 60`   `late_fine(100) → 500`

```python
def late_fine(days_late):
    return min(max(days_late - 3, 0) * 10, 500)
```

```python
check(late_fine, [((-5,), 0), ((0,), 0), ((3,), 0), ((4,), 10), ((9,), 60), ((53,), 500), ((100,), 500)])
```

### Q9 · Valid plan?

Plans are `basic`, `pro`, `elite`. Return `True` if the text is one of them **ignoring case and surrounding spaces**.

`is_valid_plan(" Pro ") → True`   `is_valid_plan("gold") → False`

```python
def is_valid_plan(text):
    return text.strip().lower() in ("basic", "pro", "elite")
```

```python
check(is_valid_plan, [((" Pro ",), True), (("ELITE",), True), (("gold",), False), (("",), False), (("basic",), True)])
```
