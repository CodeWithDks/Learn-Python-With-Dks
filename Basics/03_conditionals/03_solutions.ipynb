## Setup

```python
import math

def check(fn, cases, label=None):
    """Run fn on every (args_tuple, expected) pair and report. Never crashes the notebook."""
    label = label or fn.__name__
    failed = 0
    for args, expected in cases:
        try:
            got = fn(*args)
        except Exception as e:
            got = f"raised {type(e).__name__}: {e}"
        same = (math.isclose(got, expected, abs_tol=1e-9)
                if isinstance(got, float) and isinstance(expected, float) else got == expected)
        if not same:
            failed += 1
            print(f"  ❌ {label}{args}\n       expected: {expected!r}\n       got     : {got!r}")
    if failed:
        print(f"❌ {label}: {failed}/{len(cases)} checks failed - keep going")
    else:
        print(f"✅ {label}: all {len(cases)} checks passed")
```

### Q1 · Membership status

Return `"Expired"` if `days_left < 0`, `"Expiring soon"` if it's **0 to 7**, otherwise `"Active"`.

`membership_status(-1) → "Expired"`   `membership_status(7) → "Expiring soon"`   `membership_status(8) → "Active"`

```python
def membership_status(days_left):
    if days_left < 0:
        return "Expired"
    elif days_left <= 7:
        return "Expiring soon"
    return "Active"
```

```python
check(membership_status, [((-1,), "Expired"), ((0,), "Expiring soon"), ((7,), "Expiring soon"), ((8,), "Active"), ((365,), "Active")])
```

### Q2 · Plan for a budget (mind the order)

Budget **< 500** → `"None"`, **500–999** → `"Basic"`, **1000–1999** → `"Pro"`, **2000+** → `"Elite"`.

`plan_for_budget(1000) → "Pro"`

```python
def plan_for_budget(budget):
    if budget >= 2000:
        return "Elite"
    elif budget >= 1000:
        return "Pro"
    elif budget >= 500:
        return "Basic"
    return "None"
```

```python
check(plan_for_budget, [((0,), "None"), ((499,), "None"), ((500,), "Basic"), ((999,), "Basic"), ((1000,), "Pro"), ((1999,), "Pro"), ((2000,), "Elite"), ((9999,), "Elite")])
```

### Q3 · Gate decision (priority order)

Return one string, **first matching rule wins**:

1. not a member → `"DENY: not a member"`
2. `days_left < 0` → `"DENY: expired"`
3. already checked in → `"DENY: already inside"`
4. `days_left <= 7` → `"ALLOW (renew soon)"`
5. otherwise → `"ALLOW"`

```python
def gate(is_member, days_left, checked_in):
    if not is_member:
        return "DENY: not a member"
    if days_left < 0:
        return "DENY: expired"
    if checked_in:
        return "DENY: already inside"
    if days_left <= 7:
        return "ALLOW (renew soon)"
    return "ALLOW"
```

```python
check(gate, [
    ((False, 10, False), "DENY: not a member"),
    ((False, -5, True), "DENY: not a member"),
    ((True, -1, True), "DENY: expired"),
    ((True, 10, True), "DENY: already inside"),
    ((True, 3, False), "ALLOW (renew soon)"),
    ((True, 90, False), "ALLOW"),
])
```

### Q5 · First non-empty (the 0 trap)

Return the **first truthy** of `a`, `b`, `c`; if none is truthy return `"N/A"`. One expression is enough.

`first_non_empty("", None, "Asha") → "Asha"`   `first_non_empty(0, [], "") → "N/A"`

```python
def first_non_empty(a, b, c):
    return a or b or c or "N/A"
```

```python
check(first_non_empty, [(("", None, "Asha"), "Asha"), ((0, [], ""), "N/A"), (("x", "y", "z"), "x"), ((0, 0, 7), 7)])
```

### Q6 · Visits message (None vs 0)

Return `"No data"` **only** when `visits` is `None`. Otherwise `"<n> visits"`. Zero is real data!

`visits_message(0) → "0 visits"`   `visits_message(None) → "No data"`

```python
def visits_message(visits):
    if visits is None:
        return "No data"
    return f"{visits} visits"
```

```python
check(visits_message, [((None,), "No data"), ((0,), "0 visits"), ((12,), "12 visits")])
```

### Q7 · Day type with match-case

Use `match`/`case`: `sat`/`sun` → `"weekend"`, `mon`…`fri` → `"weekday"`, anything else → `"invalid"`. Ignore upper/lower case.

`day_type("SUN") → "weekend"`

```python
def day_type(day):
    match day.lower():
        case "sat" | "sun":
            return "weekend"
        case "mon" | "tue" | "wed" | "thu" | "fri":
            return "weekday"
        case _:
            return "invalid"
```

```python
check(day_type, [(("SUN",), "weekend"), (("sat",), "weekend"), (("Mon",), "weekday"), (("fri",), "weekday"), (("xyz",), "invalid"), (("",), "invalid")])
```

### Q8 · BMI category

BMI = weight(kg) / height(m)². Return **"Underweight"** (< 18.5), **"Normal"** (< 25), **"Overweight"** (< 30), else **"Obese"**.

`bmi_category(70, 1.75) → "Normal"`

```python
def bmi_category(weight_kg, height_m):
    bmi = weight_kg / height_m ** 2
    if bmi < 18.5:
        return "Underweight"
    elif bmi < 25:
        return "Normal"
    elif bmi < 30:
        return "Overweight"
    return "Obese"
```

```python
check(bmi_category, [((50, 1.8), "Underweight"), ((70, 1.75), "Normal"), ((85, 1.75), "Overweight"), ((100, 1.7), "Obese"), ((50, 1.5), "Normal")])
```

### Q9 · Triangle type

Return `"invalid"` if any side is ≤ 0 **or** the two shorter sides don't add up to more than the longest. Otherwise `"equilateral"` (3 equal), `"isosceles"` (exactly 2 equal) or `"scalene"`.

`triangle_type(3, 4, 5) → "scalene"`   `triangle_type(1, 2, 3) → "invalid"`

```python
def triangle_type(a, b, c):
    x, y, z = sorted((a, b, c))
    if x <= 0 or x + y <= z:
        return "invalid"
    if a == b == c:
        return "equilateral"
    if a == b or b == c or a == c:
        return "isosceles"
    return "scalene"
```

```python
check(triangle_type, [((3, 3, 3), "equilateral"), ((3, 3, 5), "isosceles"), ((3, 4, 5), "scalene"), ((1, 2, 3), "invalid"), ((0, 4, 4), "invalid"), ((-1, 2, 2), "invalid"), ((5, 3, 3), "isosceles")])
```
