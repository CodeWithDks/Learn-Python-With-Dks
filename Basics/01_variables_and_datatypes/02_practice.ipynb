## Setup

```python
import math

def check(fn, cases, label=None):
    """Run fn on every (args_tuple, expected) pair and report. Never crashes the notebook."""
    label = label or fn.__name__
    failed = 0
    for args, expected in cases:
        try:
            got = fn(*args)
        except Exception as e:
            got = f"raised {type(e).__name__}: {e}"
        same = (math.isclose(got, expected, abs_tol=1e-9)
                if isinstance(got, float) and isinstance(expected, float) else got == expected)
        if not same:
            failed += 1
            print(f"  ❌ {label}{args}\n       expected: {expected!r}\n       got     : {got!r}")
    if failed:
        print(f"❌ {label}: {failed}/{len(cases)} checks failed - keep going")
    else:
        print(f"✅ {label}: all {len(cases)} checks passed")
```
