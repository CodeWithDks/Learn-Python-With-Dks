## Setup

```python
import math

def check(fn, cases, label=None):
    """Run fn on every (args_tuple, expected) pair and report. Never crashes the notebook."""
    label = label or fn.__name__
    failed = 0
    for args, expected in cases:
        try:
            got = fn(*args)
        except Exception as e:
            got = f"raised {type(e).__name__}: {e}"
        same = (math.isclose(got, expected, abs_tol=1e-9)
                if isinstance(got, float) and isinstance(expected, float) else got == expected)
        if not same:
            failed += 1
            print(f"  ❌ {label}{args}\n       expected: {expected!r}\n       got     : {got!r}")
    if failed:
        print(f"❌ {label}: {failed}/{len(cases)} checks failed - keep going")
    else:
        print(f"✅ {label}: all {len(cases)} checks passed")
```

### Q1 · Next year's age (convert at the edge)

The form hands you the age as **text**, e.g. `"25"`. Return the member's age next year as an **int**.

`next_year_age("25") → 26`

```python
def next_year_age(age_text):
    return int(age_text) + 1
```

```python
check(next_year_age, [(("25",), 26), (("  7 ",), 8), (("0",), 1)])
```

### Q2 · Swap without a temp variable

Return the two values swapped, as a tuple. Don't use a third variable.

`swap("pro", "basic") → ("basic", "pro")`

```python
def swap(a, b):
    a, b = b, a
    return a, b
```

```python
check(swap, [(("pro", "basic"), ("basic", "pro")), ((1, 2), (2, 1))])
```

### Q3 · Price with GST

Return `price` plus GST, **rounded to 2 decimals**. GST defaults to 18%.

`total_with_gst(1000) → 1180.0`   `total_with_gst(999.99, 5) → 1049.99`

```python
def total_with_gst(price, gst_percent=18):
    return round(price + price * gst_percent / 100, 2)
```

```python
check(total_with_gst, [((1000,), 1180.0), ((999.99, 5), 1049.99), ((0,), 0.0)])
```

### Q4 · One receipt line

Return a string: the name **left-aligned in 12 characters**, then `|`, then the amount **right-aligned in 10 characters with 2 decimals**.

`receipt_line("Rahul", 1180.5) → "Rahul       |   1180.50"`

```python
def receipt_line(name, amount):
    return f"{name:<12}|{amount:>10.2f}"
```

```python
check(receipt_line, [(("Rahul", 1180.5), "Rahul       |   1180.50"), (("Asha", 5), "Asha        |      5.00")])
```

### Q5 · How many truthy values?

Return **how many** of the three arguments are truthy (as an `int`).

`count_truthy(0, 1, "x") → 2`   `count_truthy([], None, "") → 0`

```python
def count_truthy(a, b, c):
    return bool(a) + bool(b) + bool(c)
```

```python
check(count_truthy, [((0, 1, "x"), 2), (([], None, ""), 0), (("0", [0], " "), 3)])
```

### Q6 · Equal vs identical

Return a tuple `(equal, identical)`: whether `a == b` and whether `a is b`.

The check passes two *different* lists with equal contents, and the *same* list twice.

```python
def compare(a, b):
    return (a == b, a is b)
```

```python
shared = [1, 2]
check(compare, [(([1, 2], [1, 2]), (True, False)), ((shared, shared), (True, True)), (([1], [2]), (False, False))])
```

### Q7 · Append without touching the original

Return a **new** list with `"Guest"` added at the end. The original list must stay unchanged.

`add_guest(["Asha"]) → ["Asha", "Guest"]` (and the input is still `["Asha"]`)

```python
def add_guest(members):
    new_list = members.copy()
    new_list.append("Guest")
    return new_list
```

```python
original = ["Asha", "Ravi"]
result = add_guest(original)
check(add_guest, [((["Asha", "Ravi"],), ["Asha", "Ravi", "Guest"])])
if result is not None:
    print("   original untouched:", "✅" if original == ["Asha", "Ravi"] else f"❌ it became {original}")
```

### Q8 · Build a member card

Given **text** inputs, return a dict with: `name` (stripped, title-cased), `age` (int), `plan` (upper-case), `total` (price per month × months, as float, 2 decimals) and `is_adult` (age ≥ 18).

`build_member_card(" asha rao ", "17", "pro", "3", "1500")` → `{'name': 'Asha Rao', 'age': 17, 'plan': 'PRO', 'total': 4500.0, 'is_adult': False}`

```python
def build_member_card(name, age, plan, months, price_per_month):
    age = int(age)
    return {
        "name": name.strip().title(),
        "age": age,
        "plan": plan.upper(),
        "total": round(int(months) * float(price_per_month), 2),
        "is_adult": age >= 18,
    }
```

```python
check(build_member_card, [
    ((" asha rao ", "17", "pro", "3", "1500"), {"name": "Asha Rao", "age": 17, "plan": "PRO", "total": 4500.0, "is_adult": False}),
    (("RAVI", "30", "basic", "12", "999.50"), {"name": "Ravi", "age": 30, "plan": "BASIC", "total": 11994.0, "is_adult": True}),
])
```
